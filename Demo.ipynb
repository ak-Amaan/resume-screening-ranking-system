# Resume Screening and Ranking Demo

This notebook demonstrates the local pipeline: loading resumes, parsing, job-description parsing, embedding generation, feature engineering, training, prediction, and ranking.

```python
from pathlib import Path

from parser.pdf_extractor import extract_text_from_pdf
from parser.resume_parser import parse_resume
from parser.job_description_parser import parse_job_description
from nlp.embeddings import get_embedding
from nlp.similarity import calculate_resume_job_similarity
from ml.features import create_feature_vector
from ml.train_model import train_candidate_ranking_model
from predict_candidates import predict_candidates

resume_paths = sorted(Path('data/resumes').glob('*.pdf'))
job_path = Path('data/job_descriptions/data_scientist.txt')
resume_paths[:2], job_path
```

## Load and Parse a Resume

```python
resume_text = extract_text_from_pdf(resume_paths[0])
resume = parse_resume(resume_text)
resume.to_dict()
```

## Parse a Job Description

```python
job_description = parse_job_description(job_path.read_text(encoding='utf-8'))
job_description
```

## Generate Embeddings and Similarity

```python
embedding = get_embedding(resume_text)
similarity = calculate_resume_job_similarity(resume, job_description)
embedding.shape, similarity
```

## Feature Engineering

```python
feature_vector = create_feature_vector(resume, job_description)
feature_vector
```

## Training

```python
model, metrics = train_candidate_ranking_model()
metrics
```

## Prediction and Ranking

```python
rankings = predict_candidates(
    resume_dir='data/resumes',
    job_description_path='data/job_descriptions/data_scientist.txt',
    output_path='output/rankings.csv',
)
rankings
```
